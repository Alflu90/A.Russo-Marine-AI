```python
# pip install crewai chromadb sentence-transformers pandas --quiet

import sys
import os

os.environ["LM_STUDIO_MODEL"] = "openai/gpt-oss-20b"

from pathlib import Path

def find_project_root(marker=".gitignore"):
    path = Path.cwd()
    while path != path.parent:
        if (path / marker).exists():
            return path
        path = path.parent
    raise FileNotFoundError(f"Marker '{marker}' non trovato risalendo da {Path.cwd()}")

PROJECT_ROOT = find_project_root()
sys.path.insert(0, str(PROJECT_ROOT / "src"))

# Decommentare e correggere se il nome del modello in LM Studio e' diverso:
# os.environ["LM_STUDIO_MODEL"] = "openai/nome-esatto-del-modello"

print("Project root:", PROJECT_ROOT)
```

```python
import requests

# Verifica che LM Studio sia raggiungibile e mostra i modelli caricati.
base_url = os.environ.get("LM_STUDIO_BASE_URL", "http://localhost:1234/v1")
resp = requests.get(f"{base_url}/models", timeout=5)
resp.raise_for_status()
for m in resp.json().get("data", []):
    print(m["id"])
```
